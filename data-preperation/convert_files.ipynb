```python
import numpy as np
import pandas as pd
import os
import openpyxl
```

```python
data_path = os.path.join(os.getcwd(), "..", "data", "training")
raw_data_path = os.path.join(data_path, "raw")
converted_data_path = os.path.join(data_path, "converted")
os.makedirs(converted_data_path, exist_ok=True)
```

```python
raw_files = os.listdir(raw_data_path)
raw_files = [file for file in raw_files if file.endswith(".xlsx")]
raw_files
```

```python
workbooks = np.empty((len(raw_files), 2), dtype=object)
for i, file in enumerate(raw_files):
    print(f"Loading {file}...")
    workbooks[i, 0] = file
    workbooks[i, 1] = openpyxl.load_workbook(os.path.join(raw_data_path, file), data_only=True)
    print(f"Loaded {file} successfully.")
```

```python
sheets = [(wb[0], wb[1].sheetnames) for wb in workbooks]
sheets
```

```python
for file, sheets_ in sheets:
    for sheet in sheets_:
        df = pd.DataFrame(workbooks[i][sheet].values)
        file_name = f"{file.split('.')[0]}_{sheet}".replace(" ", "_").replace("/", "_").replace(".", "_").replace(",", "_").replace("__", "_").lower() + ".csv"
        df.to_csv(os.path.join(converted_data_path, file_name), index=False)
```
